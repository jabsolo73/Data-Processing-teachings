```python
import pandas as pd

# Read the CSV file into a DataFrame
df = pd.read_csv("/content/logs.csv")

# Split the 'Time' column into 'Date', 'Time', and 'Day', 'Month', 'Year' columns
df['Date'] = pd.to_datetime(df['Time'])
df['Day'] = df['Date'].dt.day
df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year
df['Time'] = df['Date'].dt.time

# Save the updated DataFrame back to a CSV file
df.to_csv('updated_dataset.csv', index=False)
df.head()
```

```python
import pandas as pd

# Read the 'updated_dataset(1).csv' and 'Scores.csv' files into DataFrames
updated_df = pd.read_csv('/content/updated_dataset.csv')
scores_df = pd.read_csv("/content/scores (1).csv")

# Find common Student IDs between the two DataFrames
common_ids = set(updated_df['StudentId']).intersection(scores_df['StudentId'])

# Filter rows based on common Student IDs
updated_filtered = updated_df[updated_df['StudentId'].isin(common_ids)]
scores_filtered = scores_df[scores_df['StudentId'].isin(common_ids)]

# Save the filtered DataFrames to new CSV files
updated_filtered.to_csv('updated_dataset_filtered.csv', index=False)
scores_filtered.to_csv('scores_filtered.csv', index=False)
updated_df.head()
```

```python
import pandas as pd

# Read the data from a CSV file
df = pd.read_csv('/content/updated_dataset_filtered.csv')

# Select only the required columns
df = df[['StudentId', 'Action', 'Month', 'Grade']]

# Aggregate the data to get the total count of each action for each student and month
agg_df = df.groupby(['StudentId', 'Month', 'Action']).size().unstack(fill_value=0).reset_index()

# Merge the aggregated data with the original DataFrame to add the grades
final_df = pd.merge(df[['StudentId', 'Month', 'Grade']], agg_df, on=['StudentId', 'Month'], how='left')

# Remove duplicate rows
final_df.drop_duplicates(subset=['StudentId', 'Month'], keep='first', inplace=True)

# Save the final DataFrame to an Excel file
final_df.to_excel('output4.xlsx', index=False)
final_df.head()
```

```python
from pytube import YouTube
from moviepy.editor import *

def download_video_as_mp3(url, save_path):
    # Download video from YouTube
    yt = YouTube(url)
    video = yt.streams.filter(only_audio=True).first()
    out_file = video.download(output_path=save_path)

    # Load the downloaded video
    video_clip = VideoFileClip(out_file)

    # Set the filename for the MP3 file
    mp3_file = out_file.split('.')[0] + '.mp3'

    # Convert to MP3
    video_clip.audio.write_audiofile(mp3_file)

    # Remove the original video file
    os.remove(out_file)

    return mp3_file

# Example usage
download_video_as_mp3('https://www.youtube.com/watch?v=lBirq7-WfpQL', "C:\Users\JABEZ\Desktop")

```

```python
import os
from pytube import YouTube
from moviepy.editor import *

def sanitize_filename(filename):
    # Remove or replace characters that are invalid for file names
    valid_chars = "-_.() %s%s" % (string.ascii_letters, string.digits)
    return ''.join(c for c in filename if c in valid_chars)

def download_video_as_mp3(url, save_path):
    # Download video from YouTube
    yt = YouTube(url)
    video = yt.streams.filter(only_audio=True).first()
    out_file = video.download(output_path=save_path)

    # Sanitize the filename
    base, ext = os.path.splitext(out_file)
    sanitized_base = sanitize_filename(base)
    out_file_sanitized = sanitized_base + ext

    # Rename the file if the sanitized name is different
    if out_file_sanitized != out_file:
        os.rename(out_file, out_file_sanitized)
        out_file = out_file_sanitized

    # Load the downloaded video
    video_clip = VideoFileClip(out_file)

    # Set the filename for the MP3 file
    mp3_file = sanitized_base + '.mp3'

    # Convert to MP3
    video_clip.audio.write_audiofile(mp3_file)

    # Remove the original video file
    os.remove(out_file)

    return mp3_file

# Example usage
download_video_as_mp3("https://www.youtube.com/watch?v=lBirq7-WfpQ", "C:\Users\JABEZ\Desktop")
```
