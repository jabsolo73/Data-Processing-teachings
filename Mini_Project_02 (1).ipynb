```python
# 1. Convert 'Date' column to datetime, coercing errors
# This will turn '32-13-2024' and other invalid dates into NaT (Not a Time)
data['Date'] = pd.to_datetime(data['Date'], format='%d-%m-%Y', errors='coerce')

print(f"Number of invalid dates (NaT) after conversion: {data['Date'].isnull().sum()}")
display(data['Date'].head())
display(data['Date'].tail())
```

```python
# 2. Replace specific placeholder text and empty strings with np.nan

# Strip whitespace from all object columns first to catch ' Missing '
for col in data.select_dtypes(include='object').columns:
    data[col] = data[col].str.strip()

# Replace 'Missing' in 'Remarks' column with NaN
# data['Remarks'] = data['Remarks'].replace('Missing','NA', np.nan)

# Replace 'missing' in 'Email' column with NaN
data['Email'] = data['Email'].replace('missing', np.nan)

# Replace 'Unknown' in 'Gender' column with NaN
data['Gender'] = data['Gender'].replace('Unknown', np.nan)

# Replace 'Missing' (after stripping) in 'Status' with NaN
data['Status'] = data['Status'].replace('Missing', np.nan)

# General replacement for empty strings, 'NA', and 'None' in object columns
for col in data.select_dtypes(include='object').columns:
    data[col] = data[col].replace({'': np.nan, 'NA': np.nan, 'None': np.nan})

print("\nNull values after all replacements:")
display(data.isnull().sum())
```

```python
df["Age"] = df["Age"].abs()
df["Unit_Price"] = df["Unit_Price"].abs()
df["Sales"] = df["Sales"].abs()
```

```python
numeric_cols = df.select_dtypes(include="number").columns

for col in numeric_cols:
    plt.figure(figsize=(5,4))

    plt.boxplot(
        df[col],
        patch_artist=True,
        boxprops=dict(facecolor="lightblue"),
        medianprops=dict(color="red"),
        flierprops=dict(marker='o', markerfacecolor='orange')
    )

    plt.title(col)
    plt.grid(True)
    plt.show()
```

```python
iqr = []
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    iqr.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

iqr_df = pd.DataFrame(iqr)
print(iqr_df)
```

```python
rows_before = len(df)

for col in ["Age", "Sales"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]

rows_after = len(df)

print("Rows before:", rows_before)
print("Rows after :", rows_after)
print("Rows removed:", rows_before - rows_after)
```
