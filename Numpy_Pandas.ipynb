Creating DataFrame using pandas and performing Task on queries

```python
data = {
    "Emp_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Helen", "Ian", "Jack"],
    "Age": [25, 30, 22, 35, 28, 40, 27, 31, 29, 26],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Sales", "Finance", "IT", "Marketing", "Sales"],
    "Salary": [45000, 60000, 52000, 75000, np.nan, 68000, 49000, 72000, 58000, 61000],
    "Experience": [2, 5, 1, 8, 3, 10, 2, 6, 4, 5],
    "City": ["Chennai", "Bangalore", "Hyderabad", "Chennai", "Mumbai",
             "Delhi", "Hyderabad", "Bangalore", "Chennai", "Mumbai"],
    "Performance": ["Good", "Excellent", "Average", "Excellent", "Good",
                    "Excellent", "Average", "Good", "Excellent", "Average"]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
#1 Display all employees who belong to the IT department.
df[df["Department"]=="IT"]
```

```python
#2 Display all employees whose Age is greater than 30.
df[df["Age"] >30]
```

```python
#3 Display employees who: belong to the IT department, and have a Salary greater than 65000.
df[(df["Department"]=="IT") & (df["Salary"] > 65000)]
```

```python
#4 Display employees who are from Chennai or Mumbai
df[df["City"] .isin (["Chennai","Mumbai"])]
```

```python
#5 Find all employees whose Salary is missing (NaN).
df[df["Salary"] .notnull()]
```

```python
#6 Display employees whose department is either:HR, Finance. Sales
df[df["Department"] .isin (["HR","IT","Finance"])]
```

```python
#7 Display employees whose Experience is between 3 and 6 years (inclusive).
df[df["Experience"].between(3,6)]
```

```python
#8 Display employees who do not belong to the IT department
df.query("Department != 'IT'")
```

```python
#9 Display employees who satisfy all of the following: Performance is Excellent Salary is greater than 60000 Experience is at least 5 years
df.query("Performance == 'Excellent' and Salary > 60000 and Experience >= 5")
```

```python
#10 Display employees who satisfy either of these conditions: Age is less than 28 and Department is HR OR
# Salary is greater than 70000 and City is Bangalore Bonus Questions
df.query("Age < 28 or Department == 'HR' or Salary > 70000 or City == 'Banglore'")
```

```python
# 11 Display employees whose Name starts with the letter "A"
df[df["Name"].str.startswith("A")]
```

```python
#12 Display employees whose City contains the word "Chennai".
df[df["City"].str.contains("Chennai")]
```

```python
#13 Display employees whose Performance is not "Average".
df[df["Performance"] != 'Average']
```

```python
#14 Salary between 50000 and 70000, and Age less than 30.
df[df["Salary"].between(50000, 70000) & (df["Age"] < 30)]
```

```python
#15 belong to Finance,have Experience less than 3 years, and Performance is Average.
df.query("Department == 'Finance' and Experience < 3 and Performance == 'Average'")
```
